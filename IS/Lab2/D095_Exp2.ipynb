```python
def goaltest(node, goal):
    return node == goal
```

```python
def occursin(node, nodepairs):
    for pair in nodepairs:
        if pair[0] == node:
            return True
    return False
```

```python
def removeseen(children, openlist, closedlist):
    newchildren = []
    for child in children:
        if not occursin(child, openlist) and not occursin(child, closedlist):
            newchildren.append(child)

    return newchildren
```

```python
def makepairs(children, parent):
    pairs = []

    for child in children:
        pairs.append((child, parent))

    return pairs
```

```python
def reconstructpath(goalpair, closed):
    path = [goalpair[0]]
    parent = goalpair[1]

    while parent is not None:
        path.append(parent)

        found = False
        for pair in closed:
            if pair[0] == parent:
                parent = pair[1]
                found = True
                break

        if not found:
            break

    path.reverse()
    return path
```

```python
def dfs(graph, start, goal):

    openlist = [(start, None)]
    closedlist = []
    while openlist:
        nodepair = openlist.pop(0)
        node = nodepair[0]
        if goaltest(node, goal):
            return reconstructpath(nodepair, closedlist)
        closedlist.append(nodepair)
        children = movegen(graph, node)
        children = removeseen(children, openlist, closedlist)
        newpairs = makepairs(children, node)
        openlist = newpairs + openlist

    return "goal Not Found"
```

```python
graph = {}
print("Enter graph")
print("type END to stop")
while True:
    line = input()

    if line.upper() == "END":
        break

    node, children = line.split(":")

    if children.strip() == "":
        graph[node] = []
    else:
        graph[node] = children.split(",")

start = input("start Node : ")
goal = input("goal Node : ")
answer = dfs(graph, start, goal)
print("\nPath =", answer)
```
