```python
n = int(input("Number of vertices: "))

graph = {}

for _ in range(n):
    vertex = input("Vertex: ").strip()
    neighbors = input("Neighbors (space-separated): ").split()

    graph[vertex] = neighbors

print(graph)
```

```python
print("\nGraph:")
print("-" * 25)

for node, neighbors in graph.items():
    print(f"{node} -> {' '.join(neighbors)}")
```

```python
def movegen(state: str, graph: dict[str, list[str]]) -> list[str]:
    current = state[-1]          
    moves = []

    for neighbor in graph[current]:
        moves.append(state + neighbor)

    return moves
```

```python
state = input("Enter current state (e.g., s, sa, sad): ").strip()

moves = movegen(state, graph)

print("\nPossible moves:")
for move in moves:
    print(move)
```

```python
def goal_test(state: str, goal: str) -> bool:
    return state[-1] == goal
```

```python
start = input("\nStart vertex: ").strip()
goal = "g"
```

```python
print(f"\nStart: {start}")
print(f"Goal: {goal}")
```

```python
print(goal_test(start, goal))
```
