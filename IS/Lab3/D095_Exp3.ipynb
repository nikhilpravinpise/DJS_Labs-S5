```python
def movegen(graph, node):
    return list(graph.get(node, []))
```

```python
def goaltest(node, goal):
    return node == goal
```

```python
def occursin(node, nodepairs):
    return any(pair[0] == node for pair in nodepairs)
```

```python
def removeseen(children, openlist, closedlist):
    return [
        child
        for child in children
        if not occursin(child, openlist) and not occursin(child, closedlist)
    ]
```

```python
def makepairs(children, parent, depth):
    return [(child, parent, depth) for child in children]
```

```python
def reconstructpath(goalpair, closed):
    parent_map = {node: parent for node, parent, _ in closed}
    path = []
    current = goalpair[0]

    while current is not None:
        path.append(current)
        current = goalpair[1] if current == goalpair[0] else parent_map.get(current)

    path.reverse()
    return path
```

```python
def depth_limited_dfs(graph, start, goal, limit):
    stack = [(start, None, 0)]
    closedlist = []
    traversal = []

    while stack:
        node, parent, depth = stack.pop()
        traversal.append(node)

        if goaltest(node, goal):
            return reconstructpath((node, parent, depth), closedlist), traversal

        closedlist.append((node, parent, depth))

        if depth < limit:
            children = removeseen(movegen(graph, node), stack, closedlist)
            for child in reversed(children):
                stack.append((child, node, depth + 1))

    return None, traversal
```

```python
def dfid(graph, start, goal, maxdepth):
    for limit in range(maxdepth + 1):
        print("Current Depth Limit =", limit)
        path, traversal = depth_limited_dfs(graph, start, goal, limit)
        print("Traversal :", traversal)

        if path is not None:
            print("Goal Found")
            print("Path :", path)
            return path

    print("Goal Not Found")
    return None
```

```python
graph = {}
print("Enter Graph")
print("Type END to stop")
while True:
    entry = input()
    if entry.upper() == "END":
        break
    node, children_text = entry.split(":")
    if children_text == "":
        graph[node] = []
    else:
        graph[node] = children_text.split(",")
start = input("Start Node : ")
goal = input("Goal Node : ")
maxdepth = int(input("Depth : "))
dfid(graph, start, goal, maxdepth)
```
