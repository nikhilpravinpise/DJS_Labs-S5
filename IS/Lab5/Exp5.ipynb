```python
from itertools import combinations

def ClauseValue(clause, solution, variables):
    for literal in clause:
        if literal.startswith("~"):
            variable = literal[1:]
            value = not solution[variables.index(variable)]
        else:
            variable = literal
            value = solution[variables.index(variable)]
        if value:
            return True
    return False

def Evaluate(solution, clauses, variables):
    count = 0
    for clause in clauses:
        if ClauseValue(clause, solution, variables):
            count += 1
    return count

def GoalTest(solution, clauses, variables):
    return Evaluate(solution, clauses, variables) == len(clauses)

def MoveGen(solution, k):
    neighbours = []
    for positions in combinations(range(len(solution)), k):
        new_solution = solution[:]
        for position in positions:
            new_solution[position] = 1 - new_solution[position]
        neighbours.append(new_solution)
    return neighbours

def HillClimbing(solution, k, clauses, variables):
    while True:
        current_value = Evaluate(solution,clauses,variables)
        neighbours = MoveGen(solution, k)

        if neighbours == []:
            return solution
        best_solution = solution
        best_value = current_value

        for neighbour in neighbours:
            value = Evaluate(neighbour,clauses,variables)
            if value > best_value:
                best_value = value
                best_solution = neighbour

        if best_value <= current_value:
            return solution
        solution = best_solution
        print("  Solution:","".join(map(str, solution)),"Satisfied:",best_value,"/",len(clauses))
        if GoalTest(solution, clauses, variables):
            return solution

def VND(start, clauses, variables):
    solution = start
    k = 1
    while k <= 2:
        print("\nNeighborhood k =", k)
        old_value = Evaluate(solution,clauses,variables)
        solution = HillClimbing(solution,k,clauses,variables)
        new_value = Evaluate(solution,clauses,variables)
        if GoalTest(solution, clauses, variables):
            return solution
        if new_value > old_value:
            k = 1
        else:
            k = k + 1
    return solution

def DisplaySolution(solution, variables):
    print("\nFinal Assignment:")
    for i in range(len(variables)):
        if solution[i] == 1:
            print(variables[i], "= True")
        else:
            print(variables[i], "= False")

n = int(input("\nEnter number of variables: "))
variables = input("Enter variable names separated by space: ").split()
m = int(input("Enter number of clauses: "))
clauses = []
print("\nEnter clauses.")
print("Use space between variables.")
print("Use ~ for NOT.")
print("Example: A ~B C")

for i in range(m):
    clause = input("Clause " + str(i + 1) + ": ").split()
    clauses.append(clause)
start_string = input("\nEnter start state (example: 00000): ")
start = [int(bit)for bit in start_string]

print("\n")
print("SAT PROBLEM")
for i in range(len(clauses)):
    print("(" + " V ".join(clauses[i]) + ")",end="")
    if i != len(clauses) - 1:
        print(" ^ ", end="")
print()

print("\n")
print("START STATE")
print("Solution:","".join(map(str, start)))

print("Satisfied Clauses:",Evaluate(start, clauses, variables),"/",len(clauses))
solution = VND(start,clauses,variables)
print("\n")
print("FINAL RESULT")
print("Final Solution:","".join(map(str, solution)))
DisplaySolution(solution,variables)
print("\nSatisfied Clauses:",Evaluate(solution, clauses, variables),"/",len(clauses))
if GoalTest(solution, clauses, variables):
    print("\nSATISFIABLE")
else:
    print("\nNo satisfying solution found")
```
